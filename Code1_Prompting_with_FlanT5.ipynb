```python
from transformers import AutoModelForSeq2SeqLM, AutoTokenizer

# Load the tokenizer and model
tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-large") #Downloads and loads the tokenizer for the FLAN-T5 Large model.
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-large") #Downloads and loads the PyTorch version of the FLAN-T5 Large model.
```

## Summarisation

```python
prompt = """summarize:
Studies show that eating carrots helps improve vision.
Carrots contain beta-carotene, a substance that the body converts into
vitamin A, crucial for maintaining healthy eyesight.
""" #Downloads and loads the PyTorch version of the FLAN-T5 Large model.

inputs = tokenizer(
    prompt,
    return_tensors="pt",
    max_length=512,
    truncation=True,
    padding=True
)
# Converts the summarization prompt into numerical PyTorch tensors that the model can process.

# return_tensors="pt": Returns PyTorch tensors.
# max_length=512: Allows a maximum of 512 input tokens.
# truncation=True: Removes tokens beyond the maximum length.
# padding=True: Adds padding when needed.

outputs = model.generate(
    inputs["input_ids"],
    max_length=150,
    num_beams=5,
    early_stopping=True
)
# Asks the model to generate a summary.

# inputs["input_ids"]: The numerical representation of the prompt.
# max_length=150: Limits the generated output to 150 tokens.
# num_beams=5: Considers five possible output paths while generating.
# early_stopping=True: Stops when a suitable complete output is found.

print(tokenizer.decode(outputs[0], skip_special_tokens=True)) #Converts the generated numbers back into readable text and prints the summary.
```

## Translation

```python
# Define the translation prompt
translation_prompt = "translate English to Spanish: Cheese is delicious."

# Prepare inputs and generate outputs
translation_inputs = tokenizer(translation_prompt, return_tensors="pt", max_length=512, truncation=True, padding=True)
translation_outputs = model.generate(translation_inputs["input_ids"], max_length=40, num_beams=5, early_stopping=True)

# Decode and display the translation
print(tokenizer.decode(translation_outputs[0], skip_special_tokens=True))
```

## Q&A

```python
# Define the context and question
context_question = "The Great Wall of China is over 13,000 miles long. question: How long is the Great Wall of China?"

# Generate inputs
question_inputs = tokenizer(context_question, return_tensors="pt", max_length=512, truncation=True, padding=True)

# Generate outputs
question_outputs = model.generate(question_inputs["input_ids"], max_length=50, num_beams=5, early_stopping=True)

# Decode and print the answer
print(tokenizer.decode(question_outputs[0], skip_special_tokens=True))
```
